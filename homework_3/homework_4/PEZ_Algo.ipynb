# Implementation of the PEZ Algorithm from [Hard Prompts made easy](https://arxiv.org/pdf/2302.03668)

**Inputs**
- Model
- Vocabulary embedding
- Projection function
- Broadcast function
- Optimization steps
- Learning rate
- Dataset

**Steps**
 - Sample embeddings from vocabulary embeddings
 - For each optimization step do
    1. Retrieve current mini-batch
    2. Forward projection on sampled embeddings
    3. Calculate the gradient in regard to the projected embedding
    4. Apply the gradient on the continuous embedding
- Perform the final projection
